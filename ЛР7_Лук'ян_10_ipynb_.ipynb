### 4. Поділ даних на навчальну й тестову вибірки

Відокремте `Cured` від ознак. Поділіть дані у співвідношенні **80% / 20%** із `random_state=42`.

**Важливо:** дослідження кореляцій і викидів, а також обчислення параметрів стандартизації виконуйте **тільки на навчальній вибірці**.

### 6. Дослідження викидів на навчальних даних

Побудуйте діаграми розмаху для ознак `X_train`. За правилом **1,5 IQR** знайдіть кількість потенційних викидів у кожній ознаці.

На основному етапі **не видаляйте викиди автоматично**. За бажанням пізніше проведіть окремий експеримент із видаленням тільки з навчальної вибірки.

```python
# TODO: побудуйте boxplot для X_train
plt.figure(figsize=(13, 5))
X_train.boxplot(rot=90)
plt.title("Діаграми розмаху ознак")
plt.show()
```

```python
q1 = X_train.quantile(0.25)
q3 = X_train.quantile(0.75)
iqr = q3 - q1

lower_bound = q1 - 1.5 * iqr
upper_bound = q3 + 1.5 * iqr
outliers = ((X_train < lower_bound) | (X_train > upper_bound)).sum()

print("Кількість потенційних викидів:")
print(outliers.sort_values(ascending=False))
```

**Висновок про викиди:**  
Найбільше потенційних викидів виявлено в ознаках Witch's Brew (33), Unicorn Horn (20), Minotaur Horn (18) та Chimera Fang (13). У той же час, у ознаках Mermaid Tears і Troll Hair викиди повністю відсутні.   

Не варто автоматично видаляти всі такі рядки без додаткової перевірки, оскільки дані описують магічні інгредієнти. Екстремальні значення можуть бути не помилками виміру, а рідкісними, але дуже важливими комбінаціями, від яких залежить успіх лікування принцеси. Видалення цих рядків призведе до втрати важливої ​​інформації та може погіршити передбачувальну здатність моделі.

### 7. Кореляційний аналіз на навчальній вибірці

### 8. Стандартизація ознак

Скористайтеся `StandardScaler`. Для `X_train` — `fit_transform()`, для `X_test` — тільки `transform()`.

```python
scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

### 9. Логістична регресія

Навчіть `LogisticRegression` на стандартизованих навчальних даних. Оцініть модель на тестовій вибірці: **Accuracy**, `classification_report`.

```python
model = LogisticRegression(random_state=42)
model.fit(X_train_scaled, y_train)

y_pred = model.predict(X_test_scaled)

print("Accuracy:", accuracy_score(y_test, y_pred))
print("\nClassification Report:\n", classification_report(y_test, y_pred))
```

### 10. Матриця помилок і приклади прогнозів

Побудуйте `confusion_matrix(y_test, y_pred)` та **теплову карту** (`sns.heatmap` із `annot=True`, `fmt='d'`). Порівняйте справжні й передбачені класи для **10 випадкових** тестових спостережень.

```python
conf = confusion_matrix(y_test, y_pred)
plt.figure(figsize=(6, 4))
sns.heatmap(conf, annot=True, fmt='d', cmap='Blues')
plt.xlabel('Predicted')
plt.ylabel('Actual')
plt.title('Матриця помилок')
plt.show()
```

```python
results_df = pd.DataFrame({'Справжні класи': y_test, 'Передбачені класи': y_pred})
display(results_df.sample(10, random_state=42))
```

**Висновок щодо якості логістичної регресії:**  
Виходячи з наданої матриці помилок, модель частіше допускає помилково-негативні помилки - 26. Хибнопозитивних помилок трохи менше - 24.

Хоча метрика Accuracy дає гарне загальне уявлення про якість моделі за рахунок правильного передбачення більшості прикладів, у контексті порятунку принцеси її однієї недостатньо. Різні типи помилок мають різну ціну: помилково порахувати, що лікування не спрацює і, можливо, відмовитися від нього, може бути фатальніше, ніж помилково очікувати на успіх. У вибірці з 10 випадкових пророцтв наочно видно, що модель іноді помиляється в обидві сторони. Тому однієї метрики не достатньо.

### 11. Підбір гіперпараметрів

Задайте набори параметрів для `LogisticRegression`. Проведіть підбір через **GridSearchCV** та **RandomizedSearchCV** із `cv=3`, `scoring='accuracy'`. Порівняйте найкращі параметри та отриману якість.

*Підказка:* дослідіть параметри `C`, `solver`, `max_iter`. Підбір виконуйте на **навчальній** вибірці.

```python
param_grid = {
    'C': [0.01, 0.1, 1, 10, 100],
    'solver': ['liblinear', 'lbfgs'],
    'max_iter': [100, 300, 500]
}

param_dist = {
    'C': np.logspace(-3, 3, 20),
    'solver': ['liblinear', 'lbfgs', 'saga'],
    'max_iter': [100, 300, 500, 1000]
}
```

```python
grid_search = GridSearchCV(LogisticRegression(random_state=42), param_grid, cv=3, scoring='accuracy')
grid_search.fit(X_train_scaled, y_train)
print("Найкращі параметри (GridSearchCV):", grid_search.best_params_)
```

```python
random_search = RandomizedSearchCV(LogisticRegression(random_state=42), param_dist, cv=3, scoring='accuracy', n_iter=10, random_state=42)
random_search.fit(X_train_scaled, y_train)
print("Найкращі параметри (RandomizedSearchCV):", random_search.best_params_)
```

```python
best_lr_model = grid_search.best_estimator_

y_pred_best = best_lr_model.predict(X_test_scaled)
print("\nAccuracy (Best Model):", accuracy_score(y_test, y_pred_best))
print("\nClassification Report (Best Model):\n", classification_report(y_test, y_pred_best))
```

**Висновок про підбір параметрів:**  
Методи пошуку підібрали інші комбінації гіперпараметрів, проте якість моделі на тестовій вибірці не змінилася. Метрика Accuracy залишилася точно на такому ж рівні, як і у базової моделі - 89.3%. Це свідчить про те, що стандартні налаштування LogisticRegression від самого початку виявилися оптимальними для цих даних, і додатковий тюнінг не дав приросту якості.

### 12. Порівняння трьох моделей

Побудуйте також **метод опорних векторів (`SVC`)** і **дерево рішень (`DecisionTreeClassifier`)**. Порівняйте ці дві моделі з логістичною регресією за **Accuracy** та **F1-score** на однаковій тестовій вибірці. Оберіть найкращу модель і поясніть вибір.

```python
svc_model = SVC(random_state=42)
dt_model = DecisionTreeClassifier(random_state=42)

svc_model.fit(X_train_scaled, y_train)
dt_model.fit(X_train_scaled, y_train)

y_pred_svc = svc_model.predict(X_test_scaled)
y_pred_dt = dt_model.predict(X_test_scaled)

results = pd.DataFrame({
    'Model': ['Logistic Regression (Tuned)', 'SVC', 'Decision Tree'],
    'Accuracy': [
        accuracy_score(y_test, y_pred_best),
        accuracy_score(y_test, y_pred_svc),
        accuracy_score(y_test, y_pred_dt)
    ],
    'F1-score': [
        f1_score(y_test, y_pred_best),
        f1_score(y_test, y_pred_svc),
        f1_score(y_test, y_pred_dt)
    ]
})

display(results)
```
